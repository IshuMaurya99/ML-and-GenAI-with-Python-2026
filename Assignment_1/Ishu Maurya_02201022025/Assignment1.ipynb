```python
#Find area of rectangle. 
x=float(input("Enter length of rectangle"))
y=float(input("Enter breadth of rectangle"))
area=x*y
print("Area of rectangle",area)
```

```python
#Find simple interest. 
P=float(input("Enter principal amount"))
R=float(input("Enter rate of interest"))
T=float(input("Enter time(in yesrs)"))
SI=(P*R*T)/100
print("Simple Interst",SI)
```

```python
#Convert temperature from Celsius to Fahrenheit. 
c=float(input("Enter celsius temperature"))
conversion=(c*9/5)+32
print("temperature celcius to fahrenheit",conversion)
```

```python
#Calculate average of 3 numbers. 
x=float(input("Enter number 1"))
y=float(input("Enter number2"))
z=float(input("Enter number3"))
avg=(x+y+z)/3
print("Average of three numbers",avg)
```

```python
#Find square and cube of a number. 
x=int(input("Enter any number"))
sq=x**2
cu=x**3
print("Square of a number",sq)
print("Cube of a number",cu)
```

```python
#Swap two numbers without third variable.
x=int(input("Enter number1"))
y=int(input("Enter number2"))
x,y=y,x
print("x",x)
print("y",y)
```
